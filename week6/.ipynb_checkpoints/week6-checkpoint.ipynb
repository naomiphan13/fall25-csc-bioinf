## Step 2: Align and Quantify the Sample Data to the Reference Genome
**Method:** Alevin-fry

**Installation Steps:** https://www.sc-best-practices.org/introduction/raw_data_processing.html#a-real-world-example

```python
# simpleaf needs the environment variable ALEVIN_FRY_HOME to store configuration and data.
!mkdir alevin_fry_home

import os
os.environ["ALEVIN_FRY_HOME"] = "alevin_fry_home"
os.environ["REF_DIR"] = "toy_ref_read/toy_human_ref"
```

```python
# Unset MPLBACKEND
os.environ.pop("MPLBACKEND", None)
```

### Building the Index

```python
%%bash
# make splici reference
pyroe make-splici \
${REF_DIR}/fasta/genome.fa \
${REF_DIR}/genes/genes.gtf \
90 \
splici_rl90_ref

# Index the reference
salmon index \
-t $(ls splici_rl90_ref/*\.fa) \
-i salmon_index \
-p 8
```

### Mapping and Quantification

```python
%%bash
# Collect FASTQ files
## The filenames are sorted and separated by space.
fastq_dir="toy_ref_read/toy_read_fastq"
reads1_pat="R1"
reads2_pat="R2"

reads1="$(find -L ${fastq_dir} -name "*${reads1_pat}*" -type f | sort | awk '{$1=$1;print}' | paste -sd' ')"
reads2="$(find -L ${fastq_dir} -name "*${reads2_pat}*" -type f | sort | awk '{$1=$1;print}' | paste -sd' ')"

# Mapping
## Usage: salmon alevin -i index_dir -l library_type -1 reads1_files -2 reads2_files -p num_threads -o output_dir
## The variable reads1 and reads2 defined above are passed in using ${}.
salmon alevin \
-i salmon_index \
-l ISR \
-1 ${reads1} \
-2 ${reads2} \
-p 8 \
-o salmon_alevin \
--chromiumV3 \
--sketch
```

### Cell Barcode Correction and UMI Resolution

```python
%%bash
# Cell barcode correction
alevin-fry generate-permit-list \
-u 3M-february-2018.txt \
-d fw \
-i salmon_alevin \
-o alevin_fry_gpl

# Filter mapping information
alevin-fry collate \
-i alevin_fry_gpl \
-r salmon_alevin \
-t 8

# UMI resolution + quantification
alevin-fry quant -r cr-like \
-m $(ls splici_rl90_ref/*3col.tsv) \
-i alevin_fry_gpl \
-o alevin_fry_quant \
-t 8
```

### Convert to AnnData Object

```python
import pyroe

quant_dir = 'alevin_fry_quant'
adata = pyroe.load_fry(quant_dir)
print(adata)
```

## Step 3: Perform Cell Clustering 
**Method:** Leiden Modularity Algorithm

**Module Required:** Scanpy

**Output:** the clustering plot

```python
import scanpy as sc

sc.settings.verbosity = 0
sc.settings.set_figure_params(dpi=80, facecolor="white", frameon=False)
```

## k-NN Graph and Dimension Reduction via UMAP

```python
# k-NN
sc.pp.neighbors(adata, n_pcs=30)

# Dimension Reduction
sc.tl.umap(adata)
```

## Leiden Clustering

```python
# Try different resolutions
sc.tl.leiden(adata, key_added="leiden_res0_25", resolution=0.25)
sc.tl.leiden(adata, key_added="leiden_res0_5", resolution=0.5)
sc.tl.leiden(adata, key_added="leiden_res1", resolution=1.0)
```

## Visualization

```python
sc.pl.umap(
    adata,
    color=["leiden_res0_25", "leiden_res0_5", "leiden_res1"],
    legend_loc="on data"
)
```

## Step 4: Perform Automatic Cell Annotation
**Method:** CellTypist

**Annotation:** Annotate the plot with the cell types

## Environment Setup
Filter out some deprecation and performance warinings that do not affect our code

```python
import warnings

warnings.filterwarnings("ignore", category=DeprecationWarning)

from numba.core.errors import NumbaDeprecationWarning

warnings.simplefilter("ignore", category=NumbaDeprecationWarning)
```

Load the needed modules

```python
import celltypist
from celltypist import models
```

## Data Normalization and Log-Transformation

```python
adata_celltypist = adata.copy()  # make a copy of our adata
adata.layers["counts"] = adata.X.copy()
adata_celltypist.X = adata.layers["counts"]  # set adata.X to raw counts
sc.pp.normalize_total(
    adata_celltypist, target_sum=10**4
)  # normalize to 10,000 counts per cell
sc.pp.log1p(adata_celltypist)  # log-transform
# make .X dense instead of sparse, for compatibility with celltypist:
adata_celltypist.X = adata_celltypist.X.toarray()
```

## Download the celltypist models for immune cells

```python
models.download_models(
    force_update=True, model=["Immune_All_Low.pkl", "Immune_All_High.pkl"]
)
```

## Create Low & High Models

```python
model_low = models.Model.load(model="Immune_All_Low.pkl")
model_high = models.Model.load(model="Immune_All_High.pkl")
```

## Modify var_names to fit the sample

```python
import mygene
mg = mygene.MyGeneInfo()

# 1. Extract ENSG IDs
genes = list(adata_celltypist.var_names)

# 2. Query mygene
query = mg.querymany(
    genes,
    scopes='ensembl.gene',
    fields='symbol',
    species='human'
)

# 3. Build mapping
mapping = {}
for item in query:
    old = item['query']
    new = item.get('symbol', old)
    mapping[old] = new

# 4. Apply mapping
adata_celltypist.var_names = [mapping[g] for g in genes]

# 5. Fix duplicates
adata.var_names_make_unique()
```

## Run the Models

```python
predictions_high = celltypist.annotate(
    adata_celltypist, model=model_high, majority_voting=True
)
```

## Copy the Results to the original AnnData object

```python
# Transform the predictions to adata to get the full output
predictions_high_adata = predictions_high.to_adata()

adata.obs["celltypist_cell_label_coarse"] = predictions_high_adata.obs.loc[
    adata.obs.index, "majority_voting"
]
adata.obs["celltypist_conf_score_coarse"] = predictions_high_adata.obs.loc[
    adata.obs.index, "conf_score"
]
```

## For the Finer annotations

```python
predictions_low = celltypist.annotate(
    adata_celltypist, model=model_low, majority_voting=True
)

predictions_low_adata = predictions_low.to_adata()

adata.obs["celltypist_cell_label_fine"] = predictions_low_adata.obs.loc[
    adata.obs.index, "majority_voting"
]
adata.obs["celltypist_conf_score_fine"] = predictions_low_adata.obs.loc[
    adata.obs.index, "conf_score"
]
```

## Plotting

```python
sc.pl.umap(
    adata,
    color=["celltypist_cell_label_coarse", "celltypist_conf_score_coarse"],
    frameon=False,
    sort_order=False,
    wspace=1,
)
```

```python
sc.pl.umap(
    adata,
    color=["celltypist_cell_label_fine", "celltypist_conf_score_fine"],
    frameon=False,
    sort_order=False,
    wspace=1,
)
```
